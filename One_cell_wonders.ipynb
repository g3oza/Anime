<b>How can we model his height at ANY time?</b> After launch, he decelerates at a constant rate (gravity). So his height is his average speed * t, where avg speed is the 1/2(launch speed + speed at that given time.) This is basically SUVAT equation 2: <br><br>
s = displacement (height in this case) in m <br>
u = initial velocity in m/s <br>
v = velocity at any give time in m/s <br>
a = acceleration (gravity in this case, 9.8m/s/s) <br>
t = time in s <br>

$$s = \frac{1}{2}(u + v)t$$ 

<b> What's his height at the peak of his sneeze?</b> This is when v=0 so subbing into the above:<br>
$$s = \frac{1}{2}(u + 0)t$$
$$s = \frac{1}{2}ut$$
<b> At what time is he at the peak of the sneeze?</b> We can start from speed. So we know that his speed at any given time is his launch speed minus gravity times time. Cos gravity slows him down by 9.8m/s every second. Here, a looks positive, but remember, gravity opposes the direction he's moving so it's negative on the way up! This is SUVAT equation 1<br>
$$v = u + at$$ 
Since we know acceleration and velocity at the peak, we can rearrange time in terms of u and sub it into the above equation to get our target, the height at the peak of the sneeze, in terms of just the initial velocity!
$$t = \frac{v-u}{a}$$
$$t = \frac{0-u}{-9.8}$$
$$t = \frac{u}{9.8}$$
Now subbing $t = \frac{u}{9.8}$ back into $s = \frac{1}{2}ut$
$$s = \frac{1}{2}u*\frac{u}{9.8}$$
$$s = \frac{u^2}{19.6}$$
<b>So how do we figure out his launch speed?</b> Well, we use the one time point that we know, which is when he hits this slope. We know:<br>
T=31.14-25.16=5.98s, s=the distance from the ground (approximately 9/11ths of a Katara, so around 152 * 9/11=124.36cm), and a=acceleration at the time, gravity 9.8, so we can use the third SUVAT equation to figure out u, the launch speed! 
$$s = ut + \frac{1}{2}at^2$$
$$1.2436 = u*5.98 - \frac{1}{2}*9.8*5.98^2$$
$$1.2436 = 5.98u − 175.23$$
$$u = 29.5 m/s$$
<b>Figuring out how high he went!</b> Now we sub the above inital launch velocity into $s = \frac{u^2}{19.6}$
$$s = \frac{29.5^2}{19.6}$$
$$s = 44.4m$$
So Sokka says '10 feet in the air' (about 3m) but if we go by the timing and the physics, it's more like 44.4m: a 14 storey building. So I guess Aang was right!
